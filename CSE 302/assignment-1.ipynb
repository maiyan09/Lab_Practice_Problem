```python
#Question-9: Greatest of three numbers.
num1 = 10
num2 = 20
num3 = 15

if (num1 >= num2) and (num1 >= num3):
    greatest = num1
elif (num2 >= num1) and (num2 >= num3):
    greatest = num2
else:
    greatest = num3

print("greatest number among 3 numbers is:", greatest)
```

```python
#Question-34: Accessing Dictionary Values
car = {
    "brand": "Toyota",
    "model": "Camry",
    "year": 2022,
    "color": "blue"
}
# add a new key-value pair
car["owner"] = "Fatin"
print(car)

# remove a key-value pair
car.pop("color")
print(car)

# print all key-value pairs
print(car.items())
```

### Function

```python
#Question-46: Creating a Simple Function for Addition
def add(a, b):
    return a + b

print(add(5, 7))
```

```python
#Question-47: Counting Vowels in a String
def countVowels(text):
    vowels = "aeiou"
    count = 0

    for char in text.lower():
        if char in vowels:
            count += 1

    return count

text = "Hello, World!"
print("Number of vowels:", countVowels(text))
```

```python
#Question-48: Function for factorial.
def factorial(n):
    if n < 0:
        return "Factorial is only for non-negative integers."

    result = 1
    for i in range(2, n + 1):
        result *= i

    return result

num = int(input("Enter a number: "))
print(f"Factorial of {num} = {factorial(num)}")
```

```python
#Question-49: Check if a Number is Even or Odd
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

num = int(input("Enter a number: "))
print(f"{num} is {check_even_odd(num)}.")
```

```python
#Question-50: Counting Uppercase Letters in a String
def count_uppercase(text):
    count = 0
    for char in text:
        if char.isupper():
            count += 1
    return count

sample = "Hello World! PYTHON"
print(f"Number of uppercase letters: {count_uppercase(sample)}")
```
